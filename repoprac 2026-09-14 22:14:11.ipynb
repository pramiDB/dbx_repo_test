```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.getOrCreate()
```

```python
data = [("John", "HR", "M", "1971-04-01", 32000),
        ("Michael", "Sales", "M", "1966-02-01", 42000),
        ("Robert", "Sales", "M", "1955-06-01", 65000),
        ("Maria", "Finance", "F", "1974-06-01", 79000)]
df = spark.createDataFrame(data, ["Name", "dept", "gender", "dob", "salary"])

df.display()
```

```python
dept_salary = df.groupBy("dept", "gender").agg(sum("salary").alias("total_salary"))
dept_salary.display()
```
